```python
from typing import TypedDict, List
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    name: str
    age: str
    skills: List[str]
    result: str
```

```python
def greeter(state: AgentState) -> AgentState:
    """this node is for greeting"""

    state['result'] = f"{state['name']}, welcome to the system!"
    return state


def age_node(state:AgentState) -> AgentState:
    """ This node describe the user's Age"""

    state['result'] = state['result'] + f" You are {state['age']} years old!"
    return state


def skills_node(state: AgentState) -> AgentState:
    """This node is for listing the user's skills """

    state['result'] = state['result'] + f" You have skills in : {", ".join(state['skills'])}"
    return state
```

```python
graph = StateGraph(AgentState)
graph.add_node("greeter", greeter)
graph.add_node("age_node", age_node)
graph.add_node("skills_node", skills_node)
graph.set_entry_point("greeter")
graph.set_finish_point("skills_node")
graph.add_edge("greeter", "age_node")
graph.add_edge("age_node","skills_node")

app = graph.compile()
```

```python
resultt = app.invoke({"name":"Marwa", "age":"19", "skills": ["Eating", "Sleeping", "Watching_movies", "Talking to Ayoub"]})
resultt["result"]
```
